```python
import cv2
import matplotlib.pyplot as plt
import numpy as np
import glob
import os
import pandas as pd
```

## Validating the code and also checking how we can improve this!

```python
def compare(path,d,LO,UP,block_size,C):
    fig, axes = plt.subplots(1,2, figsize=(12,8))
    img = cv2.imread(path)
    img_col = cv2.cvtColor(img, cv2.COLOR_RGB2BGR)
    axes[0].imshow(img_col)
    b,g,r = cv2.split(img)
    blur = cv2.GaussianBlur(r,(d,d),0)
    thresh_cv = cv2.adaptiveThreshold(blur,255,cv2.ADAPTIVE_THRESH_MEAN_C,cv2.THRESH_BINARY, block_size,C)
    contours, hierarchy = cv2.findContours(thresh_cv, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_NONE)
    blank = np.zeros(img.shape[:2],dtype='uint8')
    cnt_large = []
    lens_cnt_large = []
    img_l = img
    for cnts in contours:
        if (cv2.arcLength(cnts,True)> LO) and (cv2.arcLength(cnts,True)<UP):
            cnt_large.append(cnts)
            lens_cnt_large.append(cv2.arcLength(cnts,True))

    cv2.drawContours(img_l, cnt_large, -1, (255,0,0),10)
    axes[1].imshow(img_l)
    print('*'*90)
    print(path, '\nCounted number of plastics: ', len(cnt_large))
    plt.show()
```

```python
# currently can only run one size at a time
# takes in the paths of all the files
# define what size you want. Options are small, medium or large. Default is small

def csv_small_dump(files,params, size='small'):
    file_path = []
    count = []
    for i in range(len(files)):
        file_path.append(os.path.basename(files[i]))
        observer_counts = []
        for observers in params[size].keys():
            d = params[size][observers][0]
            LO = params[size][observers][1]
            UP = params[size][observers][2]
            block_size = params[size][observers][3]
            C = params[size][observers][4]

            img = cv2.imread(files[i])
            b,g,r = cv2.split(img)
            blur = cv2.GaussianBlur(r,(d,d),0)
            thresh_cv = cv2.adaptiveThreshold(blur,255,cv2.ADAPTIVE_THRESH_MEAN_C,cv2.THRESH_BINARY, block_size,C)
            contours, hierarchy = cv2.findContours(thresh_cv, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_NONE)
            cnt_large = []
            lens_cnt_large = []
            for cnts in contours:
                if (cv2.arcLength(cnts,True)> LO) and (cv2.arcLength(cnts,True)<UP):
                    cnt_large.append(cnts)
                    lens_cnt_large.append(cv2.arcLength(cnts,True))
            observer_counts.append(len(cnt_large))
        count.append(observer_counts)
    return file_path, np.array(count)
```
